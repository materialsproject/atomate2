## 1. Setup

````python
from pymatgen.core import Structure, Lattice
from atomate2.siesta.recipes import RecipeBook
from jobflow import run_locally

# Check available recipes
print("Available RecipeBook methods:")
print("\nComplete Studies:")
print("  - complete_material_study")
print("  - quick_characterization")
print("\nElectronic Structure:")
print("  - electronic_properties")
print("  - band_structure_workflow")
print("  - dos_calculation")
print("\nVibrational Properties:")
print("  - thermal_properties")
print("  - phonon_workflow")
print("  - gruneisen_workflow")
print("  - qha_workflow")
print("\nMechanical Properties:")
print("  - mechanical_properties")
print("  - elastic_constants_workflow")
print("  - eos_workflow")
print("\nCatalysis & Surfaces:")
print("  - catalysis_study")
print("  - surface_energy_workflow")
print("  - adsorption_scanning_workflow")
print("\nDefect Studies:")
print("  - complete_defect_study")
print("  - vacancy_study")
print("  - substitution_study")
print("  - antisite_study")
print("  - interstitial_study")
print("\nConvergence Testing:")
print("  - convergence_suite")
print("  - kpoints_convergence")
print("  - basis_convergence")
print("  - complete_convergence")
print("\nFor complete details: atomate2siesta-recipe list")
````

## 2. Test Structure

````python
# Silicon structure
si = Structure(
    lattice=Lattice.cubic(5.43),
    species=["Si", "Si"],
    coords=[[0, 0, 0], [0.25, 0.25, 0.25]]
)

print(f"Test structure: {si.composition}")
````

## 3. Quick Characterization

````python
# One-liner for quick material characterization
quick_flow = RecipeBook.quick_characterization(si, dry_run=True)

print(f"Quick characterization workflow: {quick_flow.name}")
print(f"Number of jobs: {len(quick_flow.jobs)}")
````

## 4. Complete Material Study

````python
# Comprehensive characterization
complete_flow = RecipeBook.complete_material_study(si, dry_run=True)

print(f"Complete study workflow: {complete_flow.name}")
print(f"Number of jobs: {len(complete_flow.jobs)}")
````

## 5. Band Structure Workflow

````python
# Electronic structure
bands_flow = RecipeBook.band_structure_workflow(si, dry_run=True)

print(f"Band structure workflow: {bands_flow.name}")
````

## 6. Phonon Workflow

````python
# Vibrational properties
phonon_flow = RecipeBook.phonon_workflow(
    si,
    supercell_matrix=[[2, 0, 0], [0, 2, 0], [0, 0, 2]],  # 2x2x2 supercell
    dry_run=True
)

print(f"Phonon workflow: {phonon_flow.name}")
print("Includes: supercell generation, force calculations, phonopy analysis")
````

## 7. Elastic Constants

````python
# Mechanical properties
elastic_flow = RecipeBook.elastic_constants_workflow(si, dry_run=True)

print(f"Elastic workflow: {elastic_flow.name}")
````

## 8. Bulk Modulus & EOS

````python
# Equation of state and bulk modulus
bulk_mod_flow = RecipeBook.eos_workflow(si, dry_run=True)

print(f"EOS and Bulk modulus workflow: {bulk_mod_flow.name}")
print("Includes: EOS calculation, Birch-Murnaghan fitting, bulk modulus extraction")
````

## 9. Custom Parameters

````python
# Recipes accept customization
custom_flow = RecipeBook.quick_characterization(
    si,
    dry_run=True,
    tier="intermediate",  # Use intermediate tier settings
)

print(f"Custom workflow with intermediate tier: {custom_flow.name}")
````

## 10. Recipe Comparison

````python
# Compare workflow sizes
recipes = {
    "Quick characterization": RecipeBook.quick_characterization(si, dry_run=True),
    "Band structure": RecipeBook.band_structure_workflow(si, dry_run=True),
    "Phonon": RecipeBook.phonon_workflow(si, supercell_matrix=[[2,0,0],[0,2,0],[0,0,2]], dry_run=True),
    "EOS and Bulk modulus": RecipeBook.eos_workflow(si, dry_run=True),
}

print("Recipe comparison:")
print("-" * 40)
for name, flow in recipes.items():
    print(f"{name:25s}: {len(flow.jobs):3d} jobs")
````
