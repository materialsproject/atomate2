## 1. Setup

```python
import matplotlib.pyplot as plt
import numpy as np
from pymatgen.core import Structure, Lattice
from pymatgen.core.surface import SlabGenerator

from atomate2.siesta.flows.surface import MultiSurfaceEnergyFlowMaker
from jobflow import run_locally

%matplotlib inline
```

## 2. Create Bulk Structure

```python
# MgO rocksalt structure
mgo = Structure(
    lattice=Lattice.cubic(4.21),
    species=["Mg", "O"],
    coords=[[0, 0, 0], [0.5, 0.5, 0.5]]
)

print(f"Bulk structure: {mgo.composition}")
print(f"Lattice parameter: {mgo.lattice.a:.3f} A")
```

## 3. Generate Surface Slabs

```python
# Generate (100) surface slab
slabgen = SlabGenerator(
    mgo,
    miller_index=(1, 0, 0),
    min_slab_size=10,  # Angstrom
    min_vacuum_size=15,  # Angstrom
)

slabs = slabgen.get_slabs()
print(f"Generated {len(slabs)} slab termination(s) for (100)")

for i, slab in enumerate(slabs):
    print(f"  Slab {i+1}: {len(slab)} atoms")
```

## 4. Multi-Surface Workflow

Use `MultiSurfaceEnergyFlowMaker` to compare multiple surfaces:

```python
# Create multi-surface energy workflow
surface_maker = MultiSurfaceEnergyFlowMaker(
    miller_indices=[(1, 0, 0), (1, 1, 0), (1, 1, 1)],
    slab_layers=5,  # Number of atomic layers
    vacuum_size=15.0,  # Vacuum spacing in Angstroms
    symmetrize=False,  # Use asymmetric slabs
    dry_run=True,
)

surface_flow = surface_maker.make(mgo)
print(f"Multi-surface workflow created: {surface_flow.name}")
print(f"Surfaces: (100), (110), (111)")
print(f"Each surface will be calculated and energies compared")
```

```python
# Run locally (dry-run mode)
responses = run_locally(surface_flow, create_folders=True)

print("Dry-run completed!")
print("Check the generated input files in the job directory.")
```
