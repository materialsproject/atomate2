## 1. Setup

```python
import matplotlib.pyplot as plt
import numpy as np
from pymatgen.core import Structure, Lattice

from atomate2.siesta.jobs.phonon import PhonopyMaker
from atomate2.siesta.jobs.core import RelaxMaker
from jobflow import run_locally, Flow

%matplotlib inline
```

## 2. Create Structure

```python
# Silicon for phonon calculations
si = Structure(
    lattice=Lattice.cubic(5.43),
    species=["Si", "Si"],
    coords=[[0, 0, 0], [0.25, 0.25, 0.25]]
)

print(f"Structure: {si.composition}")
print(f"Atoms: {len(si)}")
```

## 3. Phonon Workflow Setup

```python
# Create phonon maker
phonon_maker = PhonopyMaker(
    min_length=10.0,  # Minimum supercell length in Angstrom
    dry_run=True,
)

phonon_flow = phonon_maker.make(si)
print(f"Phonon workflow: {phonon_flow.name}")
```

## 4. Custom Supercell

```python
# Explicit supercell matrix
phonon_maker_custom = PhonopyMaker(
    supercell_matrix=[[2, 0, 0], [0, 2, 0], [0, 0, 2]],  # 2x2x2 supercell
    dry_run=True,
)

phonon_flow_custom = phonon_maker_custom.make(si)
print("Custom 2x2x2 supercell phonon workflow created")


```

```python
# Run locally (dry-run mode)
responses = run_locally(phonon_flow_custom, create_folders=True)

print("Dry-run completed!")
print("Check the generated input files in the job directory.")
```
