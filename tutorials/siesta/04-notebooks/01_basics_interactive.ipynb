## 1. Setup

```python
from pymatgen.core import Structure, Lattice
from atomate2.siesta.jobs.core import RelaxMaker, StaticMaker
from atomate2.siesta.jobs.core import BandStructureMaker, DOSMaker
from atomate2.siesta.sets.tiers import apply_tier_preset
from jobflow import run_locally, Flow
```

## 2. Create Test Structure

```python
# Silicon diamond structure
si = Structure(
    lattice=Lattice.cubic(5.43),
    species=["Si", "Si"],
    coords=[[0, 0, 0], [0.25, 0.25, 0.25]]
)

print(f"Formula: {si.composition.reduced_formula}")
print(f"Number of atoms: {len(si)}")
```

## 3. Relaxation Workflow

```python
# Create relaxation maker with tier preset
relax_maker = RelaxMaker.fixed_cell_relaxation(dry_run=True)
relax_maker = apply_tier_preset(relax_maker, "relax_standard")

# Create and run job
relax_job = relax_maker.make(si)
print(f"Created job: {relax_job.name}")
```

## 4. Band Structure Workflow

```python
# Band structure calculation
bands_maker = BandStructureMaker(dry_run=True)

bands_job = bands_maker.make(si)
print(f"Band structure workflow created: {bands_job.name}")
```

## 5. DOS Calculation

```python
# Density of states
dos_maker = DOSMaker(dry_run=True)

dos_job = dos_maker.make(si)
print(f"DOS workflow created: {dos_job.name}")
```

## 6. Combined Workflow

Chain multiple calculations together.

```python
# Create a combined workflow
combined_flow = Flow(
    jobs=[relax_job, bands_job, dos_job],
    name="Si complete characterization"
)

print(f"Combined workflow: {combined_flow.name}")
print(f"Number of jobs: {len(combined_flow.jobs)}")
```

## 4. Run the Workflow (Dry-Run)

In dry-run mode, this generates input files without running SIESTA.

```python
# Run locally (dry-run mode)
responses = run_locally(combined_flow, create_folders=True)

print("Dry-run completed!")
print("Check the generated input files in the job directory.")
```
