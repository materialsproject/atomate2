## 1. Setup

```python
import matplotlib.pyplot as plt
import numpy as np
from pymatgen.core import Structure, Lattice

from atomate2.siesta.flows.convergence import (
    MeshCutoffConvergenceFlowMaker,
    KpointsConvergenceFlowMaker,
    MeshKpointConvergenceFlowMaker,
)
from jobflow import run_locally

%matplotlib inline
```

## 2. Test Structure

```python
# Create silicon structure
si = Structure(
    lattice=Lattice.cubic(5.43),
    species=["Si", "Si"],
    coords=[[0, 0, 0], [0.25, 0.25, 0.25]]
)

print(f"Testing convergence for: {si.composition}")
```

## 3. Mesh Cutoff Convergence

Test how total energy changes with real-space grid density.

```python
# Define mesh cutoff values to test
mesh_cutoffs = [100, 150, 200, 250, 300, 350, 400]  # in Ry

# Create convergence maker
mesh_maker = MeshCutoffConvergenceFlowMaker(
    mesh_cutoffs=mesh_cutoffs,
    dry_run=True,
)

mesh_flow = mesh_maker.make(si)
print(f"Created {len(mesh_cutoffs)} mesh cutoff calculations")
```

## 4. K-points Convergence

Test how total energy changes with k-point sampling.

```python
# Define k-point grids to test
kpoint_grids = [
    [2, 2, 2],
    [4, 4, 4],
    [6, 6, 6],
    [8, 8, 8],
    [10, 10, 10],
]

# Create convergence maker
kpt_maker = KpointsConvergenceFlowMaker(
    kpoint_grids=kpoint_grids,
    dry_run=True,
)

kpt_flow = kpt_maker.make(si)
print(f"Created {len(kpoint_grids)} k-point calculations")
```

## 5. Combined Convergence

Two-stage convergence: mesh cutoff first, then k-points.

```python
# Combined convergence workflow
combined_maker = MeshKpointConvergenceFlowMaker(
    mesh_cutoffs=[150, 200, 250, 300],
    kpoint_grids=[[4, 4, 4], [6, 6, 6], [8, 8, 8]],
    dry_run=True,
)

combined_flow = combined_maker.make(si)
print("Combined convergence workflow created")
```

## 6. Visualization (Example Data)

Example of how convergence curves look.

```python
# Example convergence data (replace with actual results)
mesh_values = [100, 150, 200, 250, 300, 350, 400]
mesh_energies = [-215.32, -215.45, -215.52, -215.55, -215.56, -215.56, -215.56]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(mesh_values, mesh_energies, 'bo-', markersize=8)
ax.axhline(y=mesh_energies[-1], color='r', linestyle='--', label='Converged')
ax.set_xlabel('Mesh Cutoff (Ry)', fontsize=12)
ax.set_ylabel('Total Energy (eV)', fontsize=12)
ax.set_title('Mesh Cutoff Convergence', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
